```python
from rich.console import Console
from rich import print
console = Console()
console.print('Esse é um texto normal.')
console.print('Esse é um texto em negrito.', style='bold')
console.print('Esse é um texto sublinhado.', style='underline')
console.print('Esse é um texto em negrito e sublinhado.', style='bold underline')
console.print('Esse é um texto vermelho.', style='red')
console.print('Esse é um texto em negrito, vermelho e sublinhado.', style='bold underline red')
console.print('Esse é um texto em negrito, vermelho e sublinhado com fundo branco.', style='bold underline red on white')
console.print('[blue]Esse é um texto [red]com uma partte vermelha[/] e outra azul[/]')
```

```python
from rich.text import Text
print('Usando Text voce consegue guardar uma string e modificar ela')
text = Text('Esse é um Text do rich')
text.stylize('bold magenta', 0, 5)
console.print(text)
text.stylize('red underline', -7)
console.print(text)
text.stylize('green bold underline', 5, 14)
console.print(text)
```

```python
from rich.theme import Theme
print('Usando Theme voce consegue criar styles e salvalos')
tema = Theme({'sucesso' : 'green bold', 'falha' : 'red underline'})
console = Console(theme=tema)
console.print('Esse é o tema de sucesso.', style='sucesso')
console.print('Esse é o tema de falha!', style='falha')
console.print('Esse é o tema de [sucesso]sucesso[/] e de [falha]falha[/]')
console.print('Você tambem pode printar emojis :thumbs_up:')
```

```python
from rich.panel import Panel
console.print(Panel('Esse é um painel do rich'))
console.print(Panel('Esse é um painel com borda vermelha e titulo', title='Titulo', style='on red'))
console.print(Panel('[white]Esse é um painel com texto branco, emojis :thumbs_up: e borda verde', style='green', width=40))
```

```python
from rich.table import Table
tabela = Table(title='Tabela de preços')
tabela.add_column('Nome')
tabela.add_column('Preço')
tabela.add_row('Lápis', '2.50')
tabela.add_row('Borracha', '1.50')
print(tabela)
```
