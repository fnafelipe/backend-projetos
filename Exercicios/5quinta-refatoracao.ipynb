```python
def processar_pedido(pedido):
    total = 0
    for item in pedido['itens']:
        if item['categoria'] == 'eletronico':
            preco = item['preco'] * item['quantidade']
            if item['quantidade'] > 5:
                preco = preco * 0.9
            total += preco
        elif item['categoria'] == 'alimento':
            preco = item['preco'] * item['quantidade']
            if item['quantidade'] > 10:
                preco = preco * 0.85
            total += preco
        else:
            preco = item['preco'] * item['quantidade']
            total += preco
    
    if pedido['cliente']['tipo'] == 'vip':
        total = total * 0.95
    
    if total > 500:
        frete = 0
    elif total > 200:
        frete = 15
    else:
        frete = 30
    
    total_final = total + frete
    
    print(f"Cliente: {pedido['cliente']['nome']}")
    print(f"Total dos itens: {total}")
    print(f"Frete: {frete}")
    print(f"Total final: {total_final}")
    
    with open('pedidos.txt', 'a') as f:
        f.write(f"{pedido['cliente']['nome']},{total_final}\n")
    
    return total_final
```
